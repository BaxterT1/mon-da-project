```python
cheese = ['Brie', 'Mozzarella', 'Feta', 'Edam', 'Gouda', 'Camembert', 'Gruyere', 'Halloumi', 'Gorgonzola']
```

```python
cheese.append('Blue')
print(cheese)
print(len(cheese))
```

```python
cheese.insert(1, 'Swiss')
print(cheese)
print(len(cheese))
```

```python
new_cheese = ['Parmesan']

cheese.extend(new_cheese)

print(cheese)
```

```python
print(len(cheese))
```

```python
cheese.remove('Swiss')
```

```python
print(cheese)
```

```python
cheese.pop(8)

print(cheese)
```

```python
cheese.pop()
print(cheese)
```

```python
fruit_1 = ['apple', 'strawberry', 'pear', 'raspberry']
fruit_2 = ['apricot', 'peach', 'fig', 'plum', 'banana']

print(fruit_1)
print(fruit_2)
```

```python
fruit = fruit_1 + fruit_2

print(fruit)
```

```python
fruit.sort()
```

```python
print(fruit)
```

```python
fruit.sort(reverse = True)
```

```python
print(fruit)
```
