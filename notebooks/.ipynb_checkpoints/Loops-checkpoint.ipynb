```python
# While loop

counter = 0

while counter < 5:
    print(counter)
    counter = counter+1
print("This will print once")
```

```python
# For loop

for i in range(5):
    print("Hello world : ", i)
```

```python
# For loop in a list
for i in [10, 100, 1000]:
    print("hello world:", i)
```

```python
# Total and average from a list of monthly counts. 
monthly_presentations = [612, 598, 634, 651, 640, 688] 
total = 0 
for month_count in monthly_presentations:
    print(month_count)
    total += month_count 
print(total)

average_presentations = total / len(monthly_presentations) 
above_average = 0 

for month_count in monthly_presentations: 
    if month_count > average_presentations: 
        above_average += 1 
print(above_average) 
```

```python
#Looping through dictionary

hospitals = {'H0301': 'Royal Melbourne Hospital', 'H0417': 'Monash Medical Centre'} 
for code, name in hospitals.items(): 
    print(f"{code}: {name}") 
```
