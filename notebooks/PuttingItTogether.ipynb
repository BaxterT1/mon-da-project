```python
# Commission per property, AUD. 
commissions = {'A': 5000, 'B': 3400, 'C': 6500, 'D': 4000, 'E': 1500} 
prices = {'A': 45000, 'B': 23400, 'C': 67000, 'D': 34600, 'E': 12900} 

# Flag properties where commission exceeds 10% of the sales price. 
for property_id, commission in commissions.items(): 
    rate = commission / prices[property_id] 
    if rate > 0.10: 
        print(f"Property {property_id}: commission rate {rate:.1%} - review") 
    else: 
        print(f"Property {property_id}: commission rate {rate:.1%}") 

def flag_high_commissions(commissions, prices, threshold=0.10): 
    """Return a list of property IDs whose commission rate exceeds threshold.""" 
    flagged = [] 
    for property_id, commission in commissions.items(): 
        if commission / prices[property_id] > threshold: 
            flagged.append(property_id) 
    return flagged 

print(flag_high_commissions(commissions, prices)) 
print(flag_high_commissions(commissions, prices, threshold=0.12)) 
```
